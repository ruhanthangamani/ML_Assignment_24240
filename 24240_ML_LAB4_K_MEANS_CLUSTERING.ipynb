## 1. Import required libraries

```python
import os

# Prevent the Windows joblib/loky physical-core detection warning.
# This does not affect the K-Means algorithm itself.
os.environ["LOKY_MAX_CPU_COUNT"] = "4"

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
```

## 2. Load the Old Faithful dataset

```python
url = "https://raw.githubusercontent.com/vincentarelbundock/Rdatasets/master/csv/datasets/faithful.csv"

df = pd.read_csv(url)
df = df[["eruptions", "waiting"]]

print("Dataset loaded successfully.")
print("\nFirst 5 rows:")
display(df.head())

print("\nDataset Shape:", df.shape)
```

## 3. Check dataset information

```python
print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDescriptive Statistics:")
display(df.describe())
```

## 4. Visualize the original dataset

```python
plt.figure(figsize=(8, 6))
plt.scatter(df["eruptions"], df["waiting"], s=50)
plt.xlabel("Eruption Duration (minutes)")
plt.ylabel("Waiting Time (minutes)")
plt.title("Old Faithful Geyser Dataset")
plt.grid(True, alpha=0.3)
plt.show()
```

## 5. Standardize the data

```python
X = df[["eruptions", "waiting"]]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Standardized data (first 5 rows):")
print(X_scaled[:5])
```

## 6. Apply K-Means clustering

```python
kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_scaled)
df["Cluster"] = clusters

print("K-Means clustering completed successfully.")
print("Number of clusters:", kmeans.n_clusters)
```

## 7. Display cluster centers

```python
# K-Means calculates centers in standardized space, so convert them
# back to the original eruption/waiting units for interpretation.
centers_scaled = kmeans.cluster_centers_
centers_original = scaler.inverse_transform(centers_scaled)

centers_df = pd.DataFrame(
    centers_original,
    columns=["eruptions", "waiting"]
)

print("Cluster Centers:")
display(centers_df)
```

## 8. Number of observations in each cluster

```python
print("Number of observations in each cluster:")
print(df["Cluster"].value_counts().sort_index())
```

## 9. Visualize the K-Means clusters and centroids

```python
plt.figure(figsize=(9, 6))

for cluster in sorted(df["Cluster"].unique()):
    cluster_data = df[df["Cluster"] == cluster]
    plt.scatter(
        cluster_data["eruptions"],
        cluster_data["waiting"],
        s=50,
        label=f"Cluster {cluster}"
    )

plt.scatter(
    centers_original[:, 0],
    centers_original[:, 1],
    s=200,
    marker="X",
    label="Centroids"
)

plt.xlabel("Eruption Duration (minutes)")
plt.ylabel("Waiting Time (minutes)")
plt.title("K-Means Clustering of Old Faithful Eruptions")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 10. Calculate the silhouette score

```python
silhouette = silhouette_score(X_scaled, clusters)
print(f"Silhouette Score: {silhouette:.4f}")
```

## 11. Analyze the clusters

```python
cluster_summary = df.groupby("Cluster")[["eruptions", "waiting"]].mean()

print("Average Eruption Duration and Waiting Time by Cluster:")
display(cluster_summary)
```

## 12. Elbow method – compare different values of K

```python
inertia = []
k_values = range(1, 7)

for k in k_values:
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    model.fit(X_scaled)
    inertia.append(model.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(k_values, inertia, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Selecting K")
plt.xticks(list(k_values))
plt.grid(True, alpha=0.3)
plt.show()
```

## 13. Display the final clustered dataset

```python
print("Final Dataset with Cluster Labels:")
display(df.head(10))
```

## Conclusion

K-Means clustering was applied to the Old Faithful geyser dataset using eruption duration and waiting time. The two features were standardized before clustering. Two clusters were created, their centroids were converted back to the original units for interpretation, and the clustering result was evaluated using the silhouette score.
